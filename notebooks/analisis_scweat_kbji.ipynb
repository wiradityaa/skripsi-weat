### Normalisasi dan tokenisasi yang digunakan

Urutan: Unicode NFKC → huruf kecil → hapus tanda petik tunggal/ganda lurus dan melengkung → karakter nonprinting `Cc/Cf` menjadi spasi → pembatas `(`, `)`, `/`, `,`, `.` menjadi spasi → rapikan whitespace → tokenisasi berdasarkan spasi. Isi kurung dan kedua sisi slash tetap digunakan. Tanda hubung dan punctuation lain tetap dipertahankan. Ini keputusan teknis operasional yang didokumentasikan, terutama perlakuan kurung/slash; tidak ada stemming, stopword removal, atau substitusi sinonim.

```python
normalized = []
for name in source['occupation_name_raw']:
    text = unicodedata.normalize('NFKC', name).lower()
    text = text.translate(str.maketrans('', '', CONFIG['preprocessing']['quotes_deleted']))
    text = ''.join(' ' if unicodedata.category(ch) in {'Cc', 'Cf'} else ch for ch in text)
    text = text.translate(str.maketrans({ch: ' ' for ch in CONFIG['preprocessing']['separators_to_space']}))
    normalized.append(re.sub(r'\s+', ' ', text).strip())
source['occupation_text'] = normalized
source['tokens'] = source['occupation_text'].str.split()
source['token_count'] = source['tokens'].str.len()
source['raw_token_count'] = source['occupation_name_raw'].str.split().str.len()
source['text_changed'] = source['occupation_name_raw'].ne(source['occupation_text'])
source['token_count_changed'] = source['raw_token_count'].ne(source['token_count'])
source['valid_code_recomputed'] = source['occupation_code'].str.fullmatch(r'2\d{3}\.\d{2}')
source['duplicate_code_recomputed'] = source['occupation_code'].duplicated(keep=False)
source['residual_recomputed'] = source['occupation_code'].str.endswith('.99') | source['occupation_text'].str.contains(r'lainnya|ybdi|ytdl', regex=True)
source['gender_leakage_terms'] = source['tokens'].map(lambda tokens: sorted(set(tokens) & set(TERMS)))
source['gender_leakage_recomputed'] = source['gender_leakage_terms'].str.len().gt(0)
source['hierarchy_complete_recomputed'] = source[['submajor_group_name', 'group_name', 'subgroup_name']].ne('').all(axis=1)
source['hierarchy_prefix_recomputed'] = (
    source['submajor_group_code'].eq(source['occupation_code'].str[:2]) &
    source['group_code'].eq(source['occupation_code'].str[:3]) &
    source['subgroup_code'].eq(source['occupation_code'].str[:4]) &
    source['submajor_group_code'].isin(CONFIG['group_order']) & source['major_group_code'].eq('2'))
source['include_analysis'] = (source['valid_code_recomputed'] & ~source['duplicate_code_recomputed'] &
    ~source['residual_recomputed'] & ~source['gender_leakage_recomputed'] &
    source['occupation_text'].ne('') & source['hierarchy_complete_recomputed'] & source['hierarchy_prefix_recomputed'])
source['exclusion_reason'] = [
    '; '.join(reason for reason, flag in [('kode tidak valid', not row.valid_code_recomputed),
        ('kode duplikat', row.duplicate_code_recomputed), ('kategori residual', row.residual_recomputed),
        ('memuat atribut gender', row.gender_leakage_recomputed), ('nama kosong', row.occupation_text == ''),
        ('hierarki tidak lengkap/sesuai', not (row.hierarchy_complete_recomputed and row.hierarchy_prefix_recomputed))] if flag)
    for row in source.itertuples()]
assert source['include_analysis'].eq(source['include_analysis_source']).all(), 'Seleksi baru berbeda; periksa audit sebelum mengukur.'
assert len(source) == 501 and source['residual_recomputed'].sum() == 85 and source['gender_leakage_recomputed'].sum() == 1
data = source.loc[source['include_analysis']].copy().sort_values('occupation_code').reset_index(drop=True)
assert len(data) == 415 and data['occupation_code'].nunique() == 415
assert data['submajor_group_code'].value_counts().sort_index().to_dict() == CONFIG['expected_group_counts']
assert data['occupation_code'].map(lambda x: isinstance(x, str)).all()
assert not data['residual_recomputed'].any() and not data['gender_leakage_recomputed'].any()
assert data['occupation_text'].ne('').all() and data['token_count'].gt(0).all()
audit_preparasi = source[['occupation_code', 'occupation_name_raw', 'occupation_text', 'tokens',
    'raw_token_count', 'token_count', 'text_changed', 'token_count_changed', 'submajor_group_code',
    'source_document', 'source_page_pdf', 'source_page_printed', 'include_analysis_source',
    'include_analysis', 'exclusion_reason', 'residual_recomputed', 'gender_leakage_recomputed', 'gender_leakage_terms']].copy()
token_composition = pd.DataFrame({'kategori': ['satu token', 'multikata'],
    'acuan_proposal': [42, 373], 'aktual': [(data['token_count'] == 1).sum(), (data['token_count'] > 1).sum()]})
token_composition['selisih'] = token_composition['aktual'] - token_composition['acuan_proposal']
counts = data.groupby('submajor_group_code').size().reindex(CONFIG['group_order'])
CHECKS += [
    {'check': 'Seleksi dihitung ulang dan cocok sumber', 'observed': int(source['include_analysis'].eq(source['include_analysis_source']).sum()), 'expected': 501, 'status': True, 'notes': 'Boolean diparse eksplisit; residual dan leakage dihitung dari teks.'},
    {'check': '415 kode string unik; kelompok sesuai', 'observed': len(data), 'expected': 415, 'status': True, 'notes': json.dumps(counts.to_dict())},
    {'check': 'Nama asli terverifikasi seluruh kode CSV validasi', 'observed': len(original_check), 'expected': 501, 'status': True, 'notes': 'Hanya nama mentah; representasi/hasil lama tidak dibaca.'},
    {'check': 'Delapan pasangan dan 16 istilah unik', 'observed': len(set(TERMS)), 'expected': 16, 'status': True, 'notes': 'Urutan proposal dan pair_id kanonis.'},
]
TIMINGS['data_preparation_seconds'] = time.perf_counter() - t0
display(token_composition)
display(audit_preparasi.loc[~source['include_analysis'], ['occupation_code', 'occupation_name_raw', 'exclusion_reason']].head(12))
display(audit_preparasi.loc[source['token_count_changed'], ['occupation_code', 'occupation_name_raw', 'occupation_text', 'raw_token_count', 'token_count']])
print({'raw': len(source), 'residual': int(source['residual_recomputed'].sum()),
       'leakage': int(source['gender_leakage_recomputed'].sum()), 'final': len(data),
       'submajor_groups': source['submajor_group_code'].nunique(), 'groups': source['group_code'].nunique(),
       'subgroups': source['subgroup_code'].nunique()})
```

## 3. Provenance model, embedding, normalisasi L2, dan OOV

Identitas model dan seluruh input dihash secara streaming. Model statis resmi fastText Bahasa Indonesia menggunakan Common Crawl/Wikipedia, CBOW dengan bobot posisi, 300 dimensi ([dokumentasi resmi](https://fasttext.cc/docs/en/crawl-vectors.html)). File biner dapat membentuk vektor subkata untuk OOV. Model pembanding di repository tidak dimuat.

Menurut Persamaan (2.2)–(2.3), ambil vektor mentah tiap token (`float64` sebelum agregasi), rata-rata dengan koefisien sama, lalu normalisasi L2. Token tidak dinormalisasi sebelum dirata-ratakan. Representasi ini tidak menangkap urutan atau konteks; koefisien sama tidak menjamin pengaruh geometris sama karena panjang/arah vektor berbeda. OOV dicatat dan tidak dibuang.

```python
t0 = time.perf_counter()
unique_tokens = sorted(set(itertools.chain.from_iterable(data['tokens'])) | set(TERMS))
token_vectors = {token: np.asarray(model.get_word_vector(token), dtype=np.float64) for token in unique_tokens}
token_audit = pd.DataFrame([
    {'token': token, 'in_vocabulary': model.get_word_id(token) >= 0,
     'dimension': vector.size, 'raw_norm': float(np.linalg.norm(vector)),
     'finite': bool(np.isfinite(vector).all()), 'is_gender_attribute': token in TERMS}
    for token, vector in token_vectors.items()])
bad_tokens = token_audit.loc[~token_audit['finite'] | token_audit['raw_norm'].le(0) | token_audit['dimension'].ne(300)]
display(bad_tokens)
assert bad_tokens.empty, 'Vektor token tidak valid; identitas ditampilkan, tidak diganti nol.'
target_raw = np.stack([np.stack([token_vectors[token] for token in tokens]).mean(axis=0) for tokens in data['tokens']])
target_norms = np.linalg.norm(target_raw, axis=1)
attribute_raw = np.stack([token_vectors[term] for term in TERMS])
attribute_norms = np.linalg.norm(attribute_raw, axis=1)
assert np.isfinite(target_raw).all() and np.all(target_norms > 0)
targets = target_raw / target_norms[:, None]
attributes = attribute_raw / attribute_norms[:, None]
token_vocab = token_audit.set_index('token')['in_vocabulary'].to_dict()
data['oov_tokens'] = data['tokens'].map(lambda tokens: [token for token in tokens if not token_vocab[token]])
data['oov_token_count'] = data['oov_tokens'].str.len()
data['has_oov'] = data['oov_token_count'].gt(0)
audit_vektor = data[['occupation_code', 'occupation_name_raw', 'occupation_text', 'token_count',
    'submajor_group_code', 'oov_tokens', 'oov_token_count', 'has_oov']].copy()
audit_vektor['dimension'] = targets.shape[1]
audit_vektor['raw_norm'] = target_norms
audit_vektor['unit_norm'] = np.linalg.norm(targets, axis=1)
audit_vektor['finite'] = np.isfinite(targets).all(axis=1)
attribute_audit = pd.DataFrame({'attribute': TERMS, 'gender': ['male'] * 8 + ['female'] * 8,
    'pair_id': PAIR_ORDER * 2, 'in_vocabulary': [token_vocab[t] for t in TERMS],
    'raw_norm': attribute_norms, 'unit_norm': np.linalg.norm(attributes, axis=1),
    'finite': np.isfinite(attributes).all(axis=1)})
assert np.allclose(audit_vektor['unit_norm'], 1, atol=CONFIG['atol'], rtol=CONFIG['rtol'])
assert np.allclose(attribute_audit['unit_norm'], 1, atol=CONFIG['atol'], rtol=CONFIG['rtol'])
TIMINGS['embedding_seconds'] = time.perf_counter() - t0
CHECKS.append({'check': 'Vektor 300 dimensi finite dan norma satu', 'observed': str(targets.shape),
    'expected': '415 × 300; 16 × 300', 'status': True, 'notes': 'Vektor token/target/atribut dicek sebelum dan setelah normalisasi.'})
display(token_audit.loc[~token_audit['in_vocabulary']])
display(attribute_audit)
display(audit_vektor.loc[audit_vektor['has_oov']].head(20))
fig, ax = plt.subplots(figsize=(7, 4))
oov_counts = [int((~data['has_oov']).sum()), int(data['has_oov'].sum())]
ax.bar(['Tanpa token OOV', 'Memiliki token OOV'], oov_counts, color=['#4C78A8', '#F58518'])
for i, n in enumerate(oov_counts): ax.text(i, n + 3, str(n), ha='center')
ax.set(title='Status vocabulary target', ylabel='Jumlah jabatan', ylim=(0, max(oov_counts) * 1.15))
finish_figure(fig, 'audit_oov', f"**OOV:** {oov_counts[1]} dari {len(data)} jabatan memiliki token OOV. Semua vektor target valid; fastText subword tetap digunakan dan OOV tidak diperlakukan sebagai kesalahan otomatis.")
del model
import gc
gc.collect()
```

## 4. Cosine, SC-WEAT, permutasi eksak, dan BH-FDR — RM 1

Cosine (Persamaan 2.4) adalah hasil kali titik vektor satuan. Untuk satu jabatan, SC-WEAT (2.8) memakai:

\[
d(w)=\frac{\overline{\cos(w,M)}-\overline{\cos(w,F)}}{\mathrm{std}(\{\cos(w,a):a\in M\cup F\},\mathrm{ddof}=0)}.
\]

Penyebut merupakan penyebaran 16 cosine **pada satu jabatan**, bukan penyebaran antarjabatan atau standard error. Positif berarti relatif lebih dekat ke laki-laki, negatif ke perempuan, nol tepat pada skor nol.

Statistik (2.9) adalah selisih jumlah `T=8×association_difference`. Seluruh `C(16,8)=12.870` pembagian berlabel dienumerasi (bukan pertukaran terbatas dalam pasangan). P-value eksak dua sisi (2.10) menghitung `|T_perm| ≥ |T_obs|`, termasuk ties, pengamatan dan komplemennya, dengan toleransi tetap `atol=rtol=1e-12`. Tidak ada koreksi tambah satu. BH (2.11) diterapkan sekali pada seluruh 415 jabatan. Persamaan (2.5)–(2.7) adalah latar WEAT, bukan pengganti pengukuran target tunggal.

```python
t0 = time.perf_counter()
cosines = targets @ attributes.T
assert cosines.shape == (415, 16) and np.isfinite(cosines).all()
assert np.max(np.abs(cosines)) <= 1 + CONFIG['atol']
partitions = np.array(list(itertools.combinations(range(16), 8)), dtype=np.int16)
permutation_signs = np.full((len(partitions), 16), -1.0, dtype=np.float64)
permutation_signs[np.arange(len(partitions))[:, None], partitions] = 1.0
assert len(partitions) == 12870 and len(np.unique(partitions, axis=0)) == 12870
assert np.all(permutation_signs.sum(axis=1) == 0)

def measure_scweat(cosine_values, male_indices, female_indices, signs, atol, rtol, q):
    """Hitung target tunggal ddof=0, exact two-sided p, dan BH satu keluarga."""
    denominator = np.std(cosine_values, axis=1, ddof=0)
    assert np.all(denominator > 0), 'Penyebut nol: skor tidak terdefinisi.'
    male_mean = cosine_values[:, male_indices].mean(axis=1)
    female_mean = cosine_values[:, female_indices].mean(axis=1)
    difference = male_mean - female_mean
    observed_signs = np.full(cosine_values.shape[1], -1.0)
    observed_signs[male_indices] = 1.0
    observed_index = np.flatnonzero(np.all(signs == observed_signs, axis=1)).item()
    permutation_statistics = cosine_values @ signs.T
    observed = permutation_statistics[:, observed_index]
    threshold = np.abs(observed)[:, None]
    extreme = ((np.abs(permutation_statistics) >= threshold) |
               np.isclose(np.abs(permutation_statistics), threshold, atol=atol, rtol=rtol)).sum(axis=1)
    p_raw = extreme / len(signs)
    reject, adjusted, _, _ = multipletests(p_raw, alpha=q, method='fdr_bh')
    return {'mean_cosine_male': male_mean, 'mean_cosine_female': female_mean,
            'association_difference': difference, 'std_cosine_all': denominator,
            'effect_size': difference / denominator, 'T_observed': observed,
            'extreme_count': extreme, 'p_raw': p_raw, 'p_adjusted': adjusted, 'reject_fdr': reject}

male_indices, female_indices = np.arange(8), np.arange(8, 16)
measurements = measure_scweat(cosines, male_indices, female_indices, permutation_signs,
                            CONFIG['atol'], CONFIG['rtol'], CONFIG['fdr_q'])
identity_columns = ['occupation_code', 'occupation_name_raw', 'occupation_text', 'submajor_group_code', 'submajor_group_name']
results = pd.concat([data[identity_columns], pd.DataFrame(measurements)], axis=1)
results['direction'] = np.select([results['effect_size'] > CONFIG['zero_atol'],
                                  results['effect_size'] < -CONFIG['zero_atol']], ['positif', 'negatif'], default='nol')
results['rank_baseline'] = results['effect_size'].rank(ascending=False, method='average')
results['reject_raw'] = results['p_raw'] <= CONFIG['fdr_q']
cosine_table = pd.concat([data[['occupation_code']], pd.DataFrame(cosines, columns=['cosine_' + t for t in TERMS])], axis=1)
assert np.allclose(results['T_observed'], 8 * results['association_difference'], atol=CONFIG['atol'], rtol=CONFIG['rtol'])
TIMINGS['main_scweat_seconds'] = time.perf_counter() - t0
display(results.head(10))
display(results['direction'].value_counts().reindex(['positif', 'negatif', 'nol'], fill_value=0).rename('jumlah').to_frame())
print({'jumlah_permutasi': len(partitions), 'raw_significant': int(results['reject_raw'].sum()),
       'BH_global_significant': int(results['reject_fdr'].sum())})
```

## 6. Leave-one-pair-out — RM 3

Satu pasangan dikeluarkan bersama-sama pada tiap kondisi; dua kelompok masing-masing berisi tujuh atribut. Penyebut dihitung ulang dari 14 cosine dengan `ddof=0`. Model, representasi target, dan cosine baru yang sama dipertahankan. Tidak ada p-value, FDR, atau ambang stabil/tidak stabil tambahan.

Baseline dipasangkan melalui kode. Peringkat memakai skor bertanda terbesar pada peringkat satu, ties memakai rata-rata. Pembalikan hanya positif ↔ negatif. Perubahan yang melibatkan nol dicatat terpisah sebagai `zero_transition`; nol → nol berarti arah dipertahankan.

```python
t0 = time.perf_counter()
lopo_frames, lopo_score_columns = [], []
for j, pair_id in enumerate(PAIR_ORDER):
    keep_male = np.delete(male_indices, j)
    keep_female = np.delete(female_indices, j)
    keep_all = np.concatenate([keep_male, keep_female])
    denom14 = cosines[:, keep_all].std(axis=1, ddof=0)
    assert np.all(denom14 > 0), f'Penyebut LOPO nol: {pair_id}'
    score = (cosines[:, keep_male].mean(axis=1) - cosines[:, keep_female].mean(axis=1)) / denom14
    condition = data[identity_columns].copy()
    condition['pair_id'] = pair_id
    condition['removed_male_term'] = TERMS[j]
    condition['removed_female_term'] = TERMS[j + 8]
    condition['n_attributes'] = len(keep_all)
    condition['std_cosine_lopo'] = denom14
    condition['effect_size_lopo'] = score
    condition['direction_lopo'] = np.select([score > CONFIG['zero_atol'], score < -CONFIG['zero_atol']], ['positif', 'negatif'], default='nol')
    condition['rank_lopo'] = condition['effect_size_lopo'].rank(ascending=False, method='average')
    condition = condition.merge(results[['occupation_code', 'effect_size', 'direction', 'rank_baseline']],
                                on='occupation_code', how='left', validate='one_to_one')
    condition = condition.rename(columns={'effect_size': 'effect_size_baseline', 'direction': 'direction_baseline'})
    condition['delta'] = condition['effect_size_lopo'] - condition['effect_size_baseline']
    condition['abs_delta'] = condition['delta'].abs()
    condition['direction_retained'] = condition['direction_lopo'].eq(condition['direction_baseline'])
    condition['sign_flip'] = ((condition['direction_baseline'].eq('positif') & condition['direction_lopo'].eq('negatif')) |
                              (condition['direction_baseline'].eq('negatif') & condition['direction_lopo'].eq('positif')))
    condition['zero_transition'] = (~condition['direction_retained'] &
        (condition['direction_baseline'].eq('nol') | condition['direction_lopo'].eq('nol')))
    condition['direction_transition'] = condition['direction_baseline'] + '→' + condition['direction_lopo']
    condition['rank_delta'] = condition['rank_lopo'] - condition['rank_baseline']
    lopo_frames.append(condition)
    lopo_score_columns.append(score)
lopo_detail = pd.concat(lopo_frames, ignore_index=True)
lopo_scores = np.column_stack(lopo_score_columns)
assert len(lopo_detail) == 3320 and not lopo_detail.duplicated(['occupation_code', 'pair_id']).any()
job_sensitivity = lopo_detail.groupby('occupation_code', sort=False).agg(
    n_conditions=('pair_id', 'size'), n_direction_retained=('direction_retained', 'sum'),
    proportion_direction_retained=('direction_retained', 'mean'), n_sign_flips=('sign_flip', 'sum'),
    n_zero_transitions=('zero_transition', 'sum'), min_effect_size_lopo=('effect_size_lopo', 'min'),
    max_effect_size_lopo=('effect_size_lopo', 'max'), max_abs_delta=('abs_delta', 'max')).reset_index()
max_pairs = []
for row in job_sensitivity.itertuples():
    subset = lopo_detail.loc[lopo_detail['occupation_code'].eq(row.occupation_code)]
    tied = subset.loc[np.isclose(subset['abs_delta'], row.max_abs_delta, atol=CONFIG['atol'], rtol=CONFIG['rtol']), 'pair_id'].tolist()
    max_pairs.append([pair_id for pair_id in PAIR_ORDER if pair_id in tied])
job_sensitivity['max_change_pairs'] = max_pairs
job_sensitivity = results[['occupation_code', 'occupation_name_raw', 'submajor_group_code', 'effect_size', 'direction']].merge(
    job_sensitivity, on='occupation_code', validate='one_to_one')
pair_rows = []
for pair_id, label in zip(PAIR_ORDER, PAIR_LABELS):
    subset = lopo_detail.loc[lopo_detail['pair_id'].eq(pair_id)]
    rho = float(spearmanr(subset['effect_size_baseline'], subset['effect_size_lopo']).statistic)
    pair_rows.append({'pair_id': pair_id, 'pair_label': label, 'n_occupations': len(subset),
        'mean_abs_delta': subset['abs_delta'].mean(), 'median_abs_delta': subset['abs_delta'].median(),
        'max_abs_delta': subset['abs_delta'].max(), 'n_sign_flips': int(subset['sign_flip'].sum()),
        'proportion_sign_flips': subset['sign_flip'].mean(), 'n_zero_transitions': int(subset['zero_transition'].sum()),
        'spearman_rho': rho, 'spearman_note': 'deskriptif; tanpa p-value' if np.isfinite(rho) else 'tidak terdefinisi: skor konstan'})
pair_sensitivity = pd.DataFrame(pair_rows)
pair_contributions = pd.concat([data[['occupation_code', 'occupation_name_raw', 'submajor_group_code']],
    pd.DataFrame(cosines[:, :8] - cosines[:, 8:], columns=PAIR_ORDER)], axis=1)
assert np.allclose(pair_contributions[PAIR_ORDER].mean(axis=1), results['association_difference'])
TIMINGS['lopo_seconds'] = time.perf_counter() - t0
display(lopo_detail.head(10))
display(pair_sensitivity)
display(job_sensitivity.nlargest(15, 'max_abs_delta'))
display(Markdown(f"**RM 3:** {job_sensitivity['n_direction_retained'].eq(8).sum()} jabatan mempertahankan arah pada semua kondisi; {job_sensitivity['n_sign_flips'].gt(0).sum()} mengalami setidaknya satu pembalikan. Tidak dibuat kategori stabil/tidak stabil. Kontribusi seluruh pasangan, termasuk pemuda–pemudi dan jejaka–gadis, disimpan pada tingkat jabatan."))
```

## 7. Validasi numerik independen dan invariansi yang dihitung ulang

Pemeriksaan memakai data aktual, bukan angka skor eksperimen lama. Cosine diperiksa dengan rumus umum. P-value beberapa jabatan dihitung ulang memakai loop referensi skalar yang terpisah dari perkalian matriks. Enumerator, komplemen, `ddof=0`, pertukaran label, BH independen, dan LOPO diuji. Invariansi baris/atribut **membentuk cosine baru dan menghitung ulang skor, p-value, BH, serta seluruh LOPO** sebelum merge berdasarkan kode (dan pasangan untuk LOPO).

```python
t0 = time.perf_counter()
sample_indices = np.unique(np.r_[np.linspace(0, len(data) - 1, 6, dtype=int),
    data.index[data['occupation_code'].isin(['2111.10', '2162.00', '2142.04'])].to_numpy()])
for i in sample_indices:
    for j in [0, 3, 8, 15]:
        general_cosine = 1 - cosine_distance(target_raw[i], attribute_raw[j])
        assert np.isclose(cosines[i, j], general_cosine, atol=CONFIG['atol'], rtol=CONFIG['rtol'])
CHECKS.append({'check': 'Cosine dot unit vs rumus umum', 'observed': len(sample_indices) * 4, 'expected': 'semua cocok', 'status': True, 'notes': 'Target/token mentah actual, scipy.spatial.distance.cosine.'})
assert np.array_equal(permutation_signs[::-1], -permutation_signs)
sample_perms = cosines[sample_indices] @ permutation_signs.T
assert np.allclose(sample_perms[:, ::-1], -sample_perms, atol=CONFIG['atol'], rtol=CONFIG['rtol'])
CHECKS.append({'check': 'Enumerator unik dan simetri komplemen', 'observed': len(partitions), 'expected': 12870, 'status': True, 'notes': 'Komplemen semua partisi dicek; statistik sampel berlawanan tanda.'})
reference_rows = []
for i in sample_indices:
    row = cosines[i]
    observed_ref = sum(float(row[k]) for k in range(8)) - sum(float(row[k]) for k in range(8, 16))
    mean_ref = sum(float(v) for v in row) / 16
    sd_ref = np.sqrt(sum((float(v) - mean_ref) ** 2 for v in row) / 16)
    effect_ref = ((sum(float(row[k]) for k in range(8)) / 8) - (sum(float(row[k]) for k in range(8, 16)) / 8)) / sd_ref
    extreme_ref = 0
    for combination in itertools.combinations(range(16), 8):
        selected = set(combination)
        statistic = sum(float(row[k]) for k in range(16) if k in selected) - sum(float(row[k]) for k in range(16) if k not in selected)
        extreme_ref += (abs(statistic) >= abs(observed_ref) or
                        np.isclose(abs(statistic), abs(observed_ref), atol=CONFIG['atol'], rtol=CONFIG['rtol']))
    assert extreme_ref == int(results.iloc[i]['extreme_count'])
    assert np.isclose(effect_ref, results.iloc[i]['effect_size'], atol=CONFIG['atol'], rtol=CONFIG['rtol'])
    reference_rows.append({'occupation_code': data.iloc[i]['occupation_code'], 'T_reference': observed_ref,
        'std_reference_ddof0': sd_ref, 'effect_reference': effect_ref,
        'extreme_count_reference': int(extreme_ref), 'p_reference': extreme_ref / 12870})
reference_checks = pd.DataFrame(reference_rows)
assert results['p_raw'].gt(0).all() and results['p_raw'].le(1).all()
assert (results['extreme_count'] >= 2).all() and (results['extreme_count'] % 2 == 0).all()
assert np.array_equal(results['p_raw'].to_numpy(), results['extreme_count'].to_numpy() / 12870)
assert np.allclose(results['std_cosine_all'], np.sqrt(((cosines - cosines.mean(axis=1, keepdims=True)) ** 2).sum(axis=1) / 16), atol=CONFIG['atol'], rtol=CONFIG['rtol'])
CHECKS.append({'check': 'Effect ddof=0 dan p-value referensi skalar', 'observed': len(sample_indices), 'expected': 'semua cocok', 'status': True, 'notes': '12870 pembagian dihitung ulang untuk setiap sampel, termasuk kode berekor nol.'})

sort_index = np.argsort(results['p_raw'].to_numpy(), kind='stable')
raw_sorted = results['p_raw'].to_numpy()[sort_index]
bh_reference_sorted = np.minimum(1, np.minimum.accumulate((raw_sorted * len(results) / np.arange(1, len(results) + 1))[::-1])[::-1])
bh_reference = np.empty(len(results)); bh_reference[sort_index] = bh_reference_sorted
assert np.allclose(bh_reference, results['p_adjusted'], atol=CONFIG['atol'], rtol=CONFIG['rtol'])
assert np.array_equal(bh_reference <= CONFIG['fdr_q'], results['reject_fdr'])
CHECKS.append({'check': 'BH independen global 415', 'observed': len(results), 'expected': 415, 'status': True, 'notes': 'Urut p, monotonic adjusted p, keputusan global cocok statsmodels.'})

swapped = measure_scweat(cosines, female_indices, male_indices, permutation_signs, CONFIG['atol'], CONFIG['rtol'], CONFIG['fdr_q'])
assert np.allclose(swapped['effect_size'], -results['effect_size'], atol=CONFIG['atol'], rtol=CONFIG['rtol'])
assert np.allclose(swapped['T_observed'], -results['T_observed'], atol=CONFIG['atol'], rtol=CONFIG['rtol'])
assert np.array_equal(swapped['p_raw'], results['p_raw']) and np.array_equal(swapped['reject_fdr'], results['reject_fdr'])
CHECKS.append({'check': 'Pertukaran label gender', 'observed': 415, 'expected': 'skor −baseline; p dua sisi sama', 'status': True, 'notes': 'Skor, T, p-value dan keputusan dihitung ulang.'})
display(reference_checks)
```

### Identitas dan kesesuaian baseline tersimpan

Lexical set memberi mapping 16 kolom cosine. Tabel hasil, data analisis, dan cosine dipasangkan satu-ke-satu berdasarkan `occupation_code`. Komponen baseline direkonstruksi dari 16 cosine dan dibandingkan dengan hasil utama. Tidak ada pembentukan embedding, p-value LOPO, atau koreksi FDR baru.

```python
d_results = d_tables['hasil_scweat']
d_lexicon = d_tables['lexicon_gender'].set_index('pair_id').loc[d_pair_order].reset_index()
assert len(d_tables['lexicon_gender']) == 8 and set(d_tables['lexicon_gender']['pair_id']) == set(d_pair_order)
assert len(d_lexicon) == 8 and d_lexicon['pair_id'].nunique() == 8
d_male_terms, d_female_terms = d_lexicon['male_term'].tolist(), d_lexicon['female_term'].tolist()
d_terms = d_male_terms + d_female_terms
assert len(set(d_terms)) == 16
d_pair_labels = (d_lexicon['male_term'] + '–' + d_lexicon['female_term']).tolist()
d_label_map = dict(zip(d_pair_order, d_pair_labels))
d_cosine_columns = ['cosine_' + term for term in d_terms]
d_data = d_tables['data_analisis']
for d_frame in [d_results, d_data, d_tables['cosine_atribut']]:
    assert len(d_frame) == 415 and d_frame['occupation_code'].nunique() == 415
    assert d_frame['occupation_code'].str.fullmatch(r'2\d{3}\.\d{2}').all()
    assert set(d_frame['occupation_code']) == set(d_results['occupation_code'])
d_core = d_results.merge(d_tables['cosine_atribut'], on='occupation_code', validate='one_to_one')
d_identity = d_core[['occupation_code', 'occupation_name_raw', 'submajor_group_code']].merge(
    d_data[['occupation_code', 'occupation_name_raw', 'submajor_group_code']], on='occupation_code',
    validate='one_to_one', suffixes=('_result', '_data'))
assert d_identity['occupation_name_raw_result'].eq(d_identity['occupation_name_raw_data']).all()
assert d_identity['submajor_group_code_result'].eq(d_identity['submajor_group_code_data']).all()
d_cosine_values = d_core[d_cosine_columns].to_numpy(dtype=np.float64)
assert d_cosine_values.shape == (415, 16) and np.isfinite(d_cosine_values).all()
d_male_mean = d_cosine_values[:, :8].mean(axis=1)
d_female_mean = d_cosine_values[:, 8:].mean(axis=1)
d_D = d_male_mean - d_female_mean
d_SD = d_cosine_values.std(axis=1, ddof=0)
assert (d_SD > 0).all()
d_baseline_effect = d_D / d_SD
for d_actual, d_stored in [(d_male_mean, 'mean_cosine_male'), (d_female_mean, 'mean_cosine_female'),
                          (d_D, 'association_difference'), (d_SD, 'std_cosine_all'), (d_baseline_effect, 'effect_size')]:
    assert np.allclose(d_actual, d_core[d_stored], atol=d_config['atol'], rtol=d_config['rtol']), d_stored
d_lopo_source = d_tables['lopo_detail'].copy()
assert len(d_lopo_source) == 3320 and not d_lopo_source.duplicated(['occupation_code', 'pair_id']).any()
assert set(d_lopo_source['pair_id']) == set(d_pair_order)
for d_column in ['sign_flip', 'zero_transition', 'direction_retained']:
    d_flags = d_lopo_source[d_column].astype(str).str.lower()
    assert d_flags.isin(['true', 'false']).all()
    d_lopo_source[d_column] = d_flags.map({'true': True, 'false': False})
d_group_counts = d_core.groupby('submajor_group_code').size().reindex(d_config['group_order'])
assert d_group_counts.notna().all() and d_group_counts.sum() == 415
d_group_names = d_core.groupby('submajor_group_code')['submajor_group_name'].first().to_dict()
d_checks += [
    {'check': 'Sumber lengkap dan checksum sembilan tabel', 'observed': len(d_tables), 'expected': 9, 'status': True},
    {'check': 'Identitas 415 kode, delapan pasangan dan 16 cosine', 'observed': '415 / 8 / 16', 'expected': '415 / 8 / 16', 'status': True},
    {'check': 'Rekonstruksi baseline dari CSV cosine', 'observed': float(np.max(np.abs(d_baseline_effect - d_core['effect_size']))), 'expected': 'dalam atol/rtol sumber', 'status': True},
]
display(d_core[['occupation_code', 'occupation_name_raw', 'effect_size', 'association_difference', 'std_cosine_all']].head())
```

### D1. Selisih cosine per pasangan — komponen pembilang

Untuk tiap jabatan dan pasangan, hitung `g = cosine_male − cosine_female`. Rata-rata delapan `g` harus sama dengan `D` baseline. Ini **selisih cosine per pasangan**, bukan effect size SC-WEAT pasangan tunggal. Arah selisih tidak membuktikan sebab semantik berupa frekuensi, usia, register, atau kekerabatan.

```python
d_pair_frames = []
for d_j, d_pair_id in enumerate(d_pair_order):
    d_piece = d_core[['occupation_code', 'occupation_name_raw', 'submajor_group_code', 'submajor_group_name']].copy()
    d_piece['pair_id'], d_piece['pair_label'] = d_pair_id, d_label_map[d_pair_id]
    d_piece['male_term'], d_piece['female_term'] = d_male_terms[d_j], d_female_terms[d_j]
    d_piece['cosine_male'], d_piece['cosine_female'] = d_cosine_values[:, d_j], d_cosine_values[:, d_j + 8]
    d_piece['cosine_difference'] = d_piece['cosine_male'] - d_piece['cosine_female']
    d_piece['difference_direction'] = np.select([d_piece['cosine_difference'] > 0, d_piece['cosine_difference'] < 0], ['positif', 'negatif'], default='nol')
    d_pair_frames.append(d_piece)
d_pair_detail = pd.concat(d_pair_frames, ignore_index=True)
d_pair_wide = d_pair_detail.pivot(index='occupation_code', columns='pair_id', values='cosine_difference').reindex(columns=d_pair_order)
d_contribution_source = d_tables['kontribusi_pasangan_per_jabatan'].set_index('occupation_code')
assert np.allclose(d_pair_wide, d_contribution_source.loc[d_pair_wide.index, d_pair_order], atol=d_config['atol'], rtol=d_config['rtol'])
d_mean_pair_check = d_pair_wide.mean(axis=1).rename('mean_pair_difference').reset_index().merge(
    d_results[['occupation_code', 'association_difference']], on='occupation_code', validate='one_to_one')
assert np.allclose(d_mean_pair_check['mean_pair_difference'], d_mean_pair_check['association_difference'], atol=d_config['atol'], rtol=d_config['rtol'])
d_pair_rows = []
for d_pair_id in d_pair_order:
    d_subset = d_pair_detail.loc[d_pair_detail['pair_id'].eq(d_pair_id)]
    d_values = d_subset['cosine_difference'].to_numpy()
    d_row = {'pair_id': d_pair_id, 'pair_label': d_label_map[d_pair_id], 'n_occupations': len(d_subset),
        'mean_cosine_difference': d_values.mean(), 'median_cosine_difference': np.median(d_values),
        'std_cosine_difference': d_values.std(ddof=0), 'min_cosine_difference': d_values.min(), 'max_cosine_difference': d_values.max()}
    for d_direction in ['positif', 'negatif', 'nol']:
        d_row['n_' + d_direction] = int(d_subset['difference_direction'].eq(d_direction).sum())
        d_row['proportion_' + d_direction] = float(d_subset['difference_direction'].eq(d_direction).mean())
    d_pair_rows.append(d_row)
d_pair_summary = pd.DataFrame(d_pair_rows)
d_positive_pair = d_pair_summary.sort_values(['mean_cosine_difference', 'pair_id'], ascending=[False, True]).iloc[0]
d_negative_pair = d_pair_summary.sort_values(['mean_cosine_difference', 'pair_id'], ascending=[True, True]).iloc[0]
d_opposition = d_pair_wide[[d_positive_pair['pair_id'], d_negative_pair['pair_id']]].rename(
    columns={d_positive_pair['pair_id']: 'positive_pair_difference', d_negative_pair['pair_id']: 'negative_pair_difference'}).reset_index()
d_opposition = d_opposition.loc[(d_opposition['positive_pair_difference'] > 0) & (d_opposition['negative_pair_difference'] < 0)].copy()
d_opposition['total_absolute_difference'] = d_opposition['positive_pair_difference'].abs() + d_opposition['negative_pair_difference'].abs()
d_opposition['positive_pair_id'], d_opposition['negative_pair_id'] = d_positive_pair['pair_id'], d_negative_pair['pair_id']
d_opposition = d_opposition.merge(d_core[['occupation_code', 'occupation_name_raw', 'submajor_group_code']], on='occupation_code', validate='one_to_one').sort_values(['total_absolute_difference', 'occupation_code'], ascending=[False, True])
d_summary['D1'] = {'largest_positive_mean_pair': d_positive_pair['pair_id'], 'positive_mean': float(d_positive_pair['mean_cosine_difference']),
    'largest_negative_mean_pair': d_negative_pair['pair_id'], 'negative_mean': float(d_negative_pair['mean_cosine_difference']), 'opposite_jobs': len(d_opposition)}
d_checks += [
    {'check': 'D1 3320 selisih cocok kontribusi sumber', 'observed': len(d_pair_detail), 'expected': 3320, 'status': True},
    {'check': 'D1 mean delapan selisih sama dengan D baseline', 'observed': float(np.max(np.abs(d_mean_pair_check['mean_pair_difference'] - d_mean_pair_check['association_difference']))), 'expected': 'dalam toleransi sumber', 'status': True},
]
display(d_pair_summary)
display(d_opposition.head(10))
```

```python
d_fig, d_ax = plt.subplots(figsize=(11, 6))
sns.boxplot(data=d_pair_detail, x='cosine_difference', y='pair_id', order=d_pair_order, color='#DCE7F3', ax=d_ax)
d_ax.set_yticks(range(8), labels=d_pair_labels)
d_ax.axvline(0, color='black', linestyle='--')
d_ax.set(title='D1 — Selisih cosine laki-laki dan perempuan per pasangan', xlabel='Selisih cosine per pasangan (komponen pembilang)', ylabel='Pasangan atribut')
d_figures.append(save_diagnostic_figure(d_fig, d_figure_dir, 'd1_boxplot_selisih_pasangan',
    f"Setiap pasangan memiliki {len(d_core)} selisih cosine. Mean tertinggi {d_positive_pair['pair_label']} ({d_positive_pair['mean_cosine_difference']:+.6f}); terendah {d_negative_pair['pair_label']} ({d_negative_pair['mean_cosine_difference']:+.6f}). Selisih ini bukan effect size pasangan tunggal atau bukti kausal semantik."))

d_fig, d_ax = plt.subplots(figsize=(11, 6))
d_ax.scatter(d_pair_summary['mean_cosine_difference'], np.arange(8) - 0.16, marker='o', s=60, label='Rata-rata', color='#0072B2')
d_ax.scatter(d_pair_summary['median_cosine_difference'], np.arange(8) + 0.16, marker='D', s=45, label='Median', color='#009E73')
for d_i, d_row in enumerate(d_pair_summary.itertuples()):
    d_ax.annotate(f'{d_row.mean_cosine_difference:+.4f}', (d_row.mean_cosine_difference, d_i - 0.16), xytext=(7, 0), textcoords='offset points', va='center', fontsize=8)
    d_ax.annotate(f'{d_row.median_cosine_difference:+.4f}', (d_row.median_cosine_difference, d_i + 0.16), xytext=(7, 0), textcoords='offset points', va='center', fontsize=8)
d_ax.set_yticks(range(8), labels=d_pair_labels)
d_ax.invert_yaxis()
d_ax.axvline(0, color='black', linestyle='--')
d_ax.margins(x=0.25)
d_ax.set(title='D1 — Pemusatan selisih cosine per pasangan', xlabel='Selisih cosine (positif: relatif laki-laki; negatif: relatif perempuan)')
d_ax.legend()
d_figures.append(save_diagnostic_figure(d_fig, d_figure_dir, 'd1_mean_median_selisih',
    f"Rata-rata delapan selisih merekonstruksi pembilang baseline pada seluruh {len(d_core)} jabatan. Ada {len(d_opposition)} jabatan dengan {d_positive_pair['pair_label']} positif dan {d_negative_pair['pair_label']} negatif; contoh diurutkan berdasarkan jumlah absolut kedua selisih, lalu kode."))
```

### D2. Besar skor awal dan pembalikan arah

Ringkasan dibangun berdasarkan kode dari baseline dan delapan kondisi LOPO. Label kejadian hanya **pernah membalik arah** atau **tidak pernah membalik arah**. Tidak ada cutoff netralitas atau kategori valid/tidak valid. Jumlah jabatan unik dibedakan dari jumlah kejadian jabatan–pasangan. Jitter seed sumber hanya menggeser tampilan jumlah pembalikan, bukan nilai analisis.

### D3. Pembilang, penyebut, dan perubahan effect size

Komponen baseline dan LOPO dihitung ulang dari cosine: dua mean, selisih `D`, SD populasi, dan `D/SD`. Untuk pasangan yang dikeluarkan, identitas aljabarnya adalah `D_LOPO = (8 D_baseline − g_j) / 7`. Penyebut tetap dihitung ulang dari **14** cosine; bukan memakai SD baseline.

Selama SD positif, **tanda D menentukan arah effect size**. Perubahan SD memengaruhi besar skor, tetapi tidak membalik tanda. Tidak dihitung persentase pembagian pengaruh kausal atau persentase perubahan effect size dekat nol.

```python
d_baseline_components = d_core[['occupation_code', 'occupation_name_raw', 'submajor_group_code', 'submajor_group_name']].copy()
d_baseline_components['n_attributes'] = 16
d_baseline_components['mean_cosine_male'], d_baseline_components['mean_cosine_female'] = d_male_mean, d_female_mean
d_baseline_components['D'], d_baseline_components['SD'], d_baseline_components['effect_size'] = d_D, d_SD, d_baseline_effect
d_baseline_components['direction'] = np.select([d_baseline_effect > 0, d_baseline_effect < 0], ['positif', 'negatif'], default='nol')
d_component_frames = []
for d_j, d_pair_id in enumerate(d_pair_order):
    d_keep_male = np.delete(np.arange(8), d_j)
    d_keep_female = np.delete(np.arange(8, 16), d_j)
    d_keep_all = np.r_[d_keep_male, d_keep_female]
    assert len(d_keep_male) == len(d_keep_female) == 7 and len(d_keep_all) == 14
    d_new_male = d_cosine_values[:, d_keep_male].mean(axis=1)
    d_new_female = d_cosine_values[:, d_keep_female].mean(axis=1)
    d_new_D = d_new_male - d_new_female
    d_new_SD = d_cosine_values[:, d_keep_all].std(axis=1, ddof=0)
    assert (d_new_SD > 0).all()
    d_new_effect = d_new_D / d_new_SD
    d_component = d_core[['occupation_code', 'occupation_name_raw', 'submajor_group_code', 'submajor_group_name']].copy()
    d_component['pair_id'], d_component['pair_label'] = d_pair_id, d_label_map[d_pair_id]
    d_component['removed_male_term'], d_component['removed_female_term'] = d_male_terms[d_j], d_female_terms[d_j]
    d_component['n_attributes_baseline'], d_component['n_attributes_lopo'] = 16, len(d_keep_all)
    d_component['mean_cosine_male_baseline'], d_component['mean_cosine_female_baseline'] = d_male_mean, d_female_mean
    d_component['D_baseline'], d_component['SD_baseline'], d_component['effect_size_baseline'] = d_D, d_SD, d_baseline_effect
    d_component['mean_cosine_male_lopo'], d_component['mean_cosine_female_lopo'] = d_new_male, d_new_female
    d_component['D_lopo'], d_component['SD_lopo'], d_component['effect_size_lopo'] = d_new_D, d_new_SD, d_new_effect
    d_component['removed_pair_cosine_difference'] = d_cosine_values[:, d_j] - d_cosine_values[:, d_j + 8]
    assert np.allclose(d_new_D, (8 * d_D - d_component['removed_pair_cosine_difference']) / 7, atol=d_config['atol'], rtol=d_config['rtol'])
    d_component['delta_D'], d_component['delta_SD'] = d_new_D - d_D, d_new_SD - d_SD
    d_component['delta_effect_size'] = d_new_effect - d_baseline_effect
    d_component['abs_delta_effect_size'] = d_component['delta_effect_size'].abs()
    d_component['direction_baseline'] = d_baseline_components['direction'].to_numpy()
    d_component['direction_lopo'] = np.select([d_new_effect > 0, d_new_effect < 0], ['positif', 'negatif'], default='nol')
    d_component['sign_flip'] = ((d_component['direction_baseline'].eq('positif') & d_component['direction_lopo'].eq('negatif')) |
                                (d_component['direction_baseline'].eq('negatif') & d_component['direction_lopo'].eq('positif')))
    d_component['zero_transition'] = (d_component['direction_baseline'].ne(d_component['direction_lopo']) &
        (d_component['direction_baseline'].eq('nol') | d_component['direction_lopo'].eq('nol')))
    d_component_frames.append(d_component)
d_components = pd.concat(d_component_frames, ignore_index=True)
d_component_check = d_components.merge(d_lopo_source, on=['occupation_code', 'pair_id'],
    validate='one_to_one', suffixes=('_reconstructed', '_source'))
assert len(d_component_check) == 3320
for d_left, d_right in [('effect_size_baseline_reconstructed', 'effect_size_baseline_source'),
    ('effect_size_lopo_reconstructed', 'effect_size_lopo_source'), ('SD_lopo', 'std_cosine_lopo'),
    ('delta_effect_size', 'delta'), ('abs_delta_effect_size', 'abs_delta')]:
    assert np.allclose(d_component_check[d_left], d_component_check[d_right], atol=d_config['atol'], rtol=d_config['rtol']), (d_left, d_right)
for d_column in ['direction_baseline', 'direction_lopo', 'sign_flip', 'zero_transition', 'removed_male_term', 'removed_female_term']:
    assert np.array_equal(d_component_check[d_column + '_reconstructed'], d_component_check[d_column + '_source']), d_column
assert np.array_equal(np.sign(d_components['D_lopo']), np.sign(d_components['effect_size_lopo']))
d_component_error = float(np.max(np.abs(d_component_check['effect_size_lopo_reconstructed'] - d_component_check['effect_size_lopo_source'])))
d_case_rows = []
for d_case_code in d_config['case_codes']:
    d_b = d_baseline_components.set_index('occupation_code').loc[d_case_code]
    d_case_rows.append({'occupation_code': d_case_code, 'occupation_name_raw': d_b['occupation_name_raw'], 'condition': 'baseline', 'pair_id': 'baseline',
        'n_attributes': 16, 'mean_cosine_male': d_b['mean_cosine_male'], 'mean_cosine_female': d_b['mean_cosine_female'],
        'D': d_b['D'], 'SD': d_b['SD'], 'effect_size': d_b['effect_size'], 'delta_D': 0.0, 'delta_SD': 0.0,
        'delta_effect_size': 0.0, 'direction': d_b['direction'], 'sign_flip': False, 'is_focus_condition': False})
    for d_pair_id in d_pair_order:
        d_c = d_components.loc[d_components['occupation_code'].eq(d_case_code) & d_components['pair_id'].eq(d_pair_id)].iloc[0]
        d_case_rows.append({'occupation_code': d_case_code, 'occupation_name_raw': d_c['occupation_name_raw'], 'condition': 'tanpa ' + d_label_map[d_pair_id],
            'pair_id': d_pair_id, 'n_attributes': 14, 'mean_cosine_male': d_c['mean_cosine_male_lopo'], 'mean_cosine_female': d_c['mean_cosine_female_lopo'],
            'D': d_c['D_lopo'], 'SD': d_c['SD_lopo'], 'effect_size': d_c['effect_size_lopo'], 'delta_D': d_c['delta_D'],
            'delta_SD': d_c['delta_SD'], 'delta_effect_size': d_c['delta_effect_size'], 'direction': d_c['direction_lopo'],
            'sign_flip': bool(d_c['sign_flip']), 'is_focus_condition': d_pair_id == d_config['case_focus_pairs'][d_case_code]})
d_cases = pd.DataFrame(d_case_rows)
d_summary['D3'] = {'max_lopo_reconstruction_error': d_component_error,
    'SD_increased_conditions': int((d_components['delta_SD'] > 0).sum()),
    'SD_decreased_conditions': int((d_components['delta_SD'] < 0).sum()),
    'cases': d_cases.loc[d_cases['pair_id'].eq('baseline') | d_cases['is_focus_condition']].to_dict(orient='records')}
d_checks += [
    {'check': 'D3 seluruh komponen LOPO cocok sumber berdasarkan kode/pasangan', 'observed': d_component_error, 'expected': '3320 baris; dalam toleransi', 'status': True},
    {'check': 'D3 SD positif, 14 atribut, identitas D dan arah', 'observed': len(d_components), 'expected': '3320; tanda D menentukan tanda skor', 'status': True},
]
display(d_components.head(8))
display(d_cases)
```

### D5. Distribusi permutasi untuk tiga contoh transparan

Contoh dipilih dari besar effect size baseline: tertinggi, terdekat median besar skor, lalu terendah. Ties dipecahkan berdasarkan kode dan kode yang sudah terpilih dikeluarkan agar tidak berulang. Ini ilustrasi ulang pengujian baseline, bukan pengujian baru.

Seluruh 12.870 statistik jumlah direkonstruksi dari 16 cosine. Pengamatan/komplemen termasuk dan kejadian ekstrem memakai toleransi sumber yang sama. Histogram menandai kejadian ekstrem, `T_observed`, dan batas dua sisi `±|T_observed|`. Adjusted p-value diambil dari BH global tersimpan, bukan dihitung ulang untuk tiga contoh atau untuk LOPO. P-value bukan probabilitas hipotesis nol benar.

```python
d_pool = d_core[['occupation_code', 'occupation_name_raw', 'effect_size', 'T_observed', 'extreme_count', 'p_raw', 'p_adjusted']].copy()
d_pool['abs_effect_size'] = d_pool['effect_size'].abs()
d_median_abs = float(d_pool['abs_effect_size'].median())
d_pool['distance_to_median_abs'] = (d_pool['abs_effect_size'] - d_median_abs).abs()
d_selected, d_selected_codes = [], []
for d_criterion, d_sort_column, d_ascending in [
    ('besar_skor_tertinggi', 'abs_effect_size', False),
    ('terdekat_median_besar_skor', 'distance_to_median_abs', True),
    ('besar_skor_terendah', 'abs_effect_size', True)]:
    d_choice = d_pool.loc[~d_pool['occupation_code'].isin(d_selected_codes)].sort_values(
        [d_sort_column, 'occupation_code'], ascending=[d_ascending, True]).iloc[0].to_dict()
    d_choice['selection_criterion'] = d_criterion
    d_choice['population_median_abs_effect_size'] = d_median_abs
    d_selected.append(d_choice)
    d_selected_codes.append(d_choice['occupation_code'])
assert len(set(d_selected_codes)) == 3
d_partition_indices = np.array(list(itertools.combinations(range(16), 8)), dtype=np.int16)
d_signs = np.full((len(d_partition_indices), 16), -1.0, dtype=np.float64)
d_signs[np.arange(len(d_partition_indices))[:, None], d_partition_indices] = 1.0
assert len(d_signs) == 12870 and len(np.unique(d_partition_indices, axis=0)) == 12870
assert np.array_equal(d_signs[::-1], -d_signs)
d_observed_signs = np.r_[np.ones(8), -np.ones(8)]
d_observed_index = np.flatnonzero(np.all(d_signs == d_observed_signs, axis=1)).item()
d_perm_frames, d_example_rows = [], []
for d_choice in d_selected:
    d_code = d_choice['occupation_code']
    d_row = d_core.set_index('occupation_code').loc[d_code, d_cosine_columns].to_numpy(dtype=np.float64)
    d_statistics = d_row @ d_signs.T
    d_observed = float(d_statistics[d_observed_index])
    d_extreme_mask = ((np.abs(d_statistics) >= abs(d_observed)) |
        np.isclose(np.abs(d_statistics), abs(d_observed), atol=d_config['atol'], rtol=d_config['rtol']))
    d_extreme_count = int(d_extreme_mask.sum())
    d_p_reconstructed = d_extreme_count / len(d_signs)
    assert np.isclose(d_observed, d_choice['T_observed'], atol=d_config['atol'], rtol=d_config['rtol'])
    assert d_extreme_count == int(d_choice['extreme_count'])
    assert np.isclose(d_p_reconstructed, d_choice['p_raw'], atol=d_config['atol'], rtol=d_config['rtol'])
    assert np.allclose(d_statistics[::-1], -d_statistics, atol=d_config['atol'], rtol=d_config['rtol'])
    # Referensi langsung sum-minus-sum, terpisah dari perkalian matriks.
    d_reference = np.array([sum(float(d_row[k]) for k in p) - sum(float(d_row[k]) for k in range(16) if k not in p)
                            for p in itertools.combinations(range(16), 8)])
    d_reference_extreme = ((np.abs(d_reference) >= abs(d_observed)) |
        np.isclose(np.abs(d_reference), abs(d_observed), atol=d_config['atol'], rtol=d_config['rtol']))
    assert np.allclose(d_reference, d_statistics, atol=d_config['atol'], rtol=d_config['rtol'])
    assert int(d_reference_extreme.sum()) == d_extreme_count
    d_example_rows.append({**d_choice, 'T_reconstructed': d_observed, 'n_partitions': len(d_signs),
        'extreme_count_reconstructed': d_extreme_count, 'p_raw_reconstructed': d_p_reconstructed,
        'p_adjusted_global': d_choice['p_adjusted']})
    d_perm_frames.append(pd.DataFrame({'occupation_code': d_code, 'selection_criterion': d_choice['selection_criterion'],
        'permutation_id': np.arange(len(d_signs)), 'T_permutation': d_statistics, 'is_extreme': d_extreme_mask}))
d_perm_examples = pd.DataFrame(d_example_rows)
d_perm_distribution = pd.concat(d_perm_frames, ignore_index=True)
d_partitions = pd.DataFrame({'permutation_id': np.arange(len(d_signs)),
    'first_group_terms': [[d_terms[k] for k in p] for p in d_partition_indices],
    'second_group_terms': [[d_terms[k] for k in range(16) if k not in p] for p in d_partition_indices]})
d_summary['D5'] = {'population_median_abs_effect_size': d_median_abs,
    'example_codes': d_selected_codes, 'examples': d_perm_examples.to_dict(orient='records')}
d_checks += [
    {'check': 'D5 tiga contoh unik dengan aturan transparan', 'observed': ', '.join(d_selected_codes), 'expected': 'maksimum / median terdekat / minimum |d|', 'status': True},
    {'check': 'D5 12870 partisi per contoh, p dan komplemen cocok', 'observed': len(d_perm_distribution), 'expected': 38610, 'status': True},
    {'check': 'D5 referensi sum-minus-sum dan adjusted p global tersimpan', 'observed': len(d_perm_examples), 'expected': 'tiga contoh; tanpa BH baru', 'status': True},
]
display(d_perm_examples)
```

```python
d_fig, d_axes = plt.subplots(1, 3, figsize=(17, 6))
d_selection_labels = {'besar_skor_tertinggi': '|d| tertinggi', 'terdekat_median_besar_skor': 'Terdekat median |d|', 'besar_skor_terendah': '|d| terendah'}
for d_ax, d_example in zip(d_axes, d_perm_examples.itertuples()):
    d_dist = d_perm_distribution.loc[d_perm_distribution['occupation_code'].eq(d_example.occupation_code)]
    d_plot_limit = float(np.abs(d_dist['T_permutation']).max())
    d_bins = np.linspace(-d_plot_limit, d_plot_limit, d_config['permutation_bins'] + 1)
    d_ax.hist([d_dist.loc[~d_dist['is_extreme'], 'T_permutation'], d_dist.loc[d_dist['is_extreme'], 'T_permutation']],
        bins=d_bins, stacked=True, color=['#A6C8E0', '#E69F00'], label=['Di luar hitungan ekstrem', 'Termasuk hitungan ekstrem'])
    d_ax.axvline(-abs(d_example.T_reconstructed), color='#666666', linestyle='--', linewidth=1.2, label='Batas ±|T pengamatan|')
    d_ax.axvline(abs(d_example.T_reconstructed), color='#666666', linestyle='--', linewidth=1.2)
    d_ax.axvline(d_example.T_reconstructed, color='black', linewidth=1.8, label='T pengamatan')
    d_ax.set(title=d_selection_labels[d_example.selection_criterion] + '\n' + d_example.occupation_code + ' — ' + fill(d_example.occupation_name_raw, 28)
        + f'\nd={d_example.effect_size:+.4f}; ekstrem={d_example.extreme_count_reconstructed}/12870\np={d_example.p_raw_reconstructed:.6f}; adjusted p={d_example.p_adjusted_global:.6f}',
        xlabel='Statistik jumlah T pada partisi atribut', ylabel='Jumlah partisi')
    d_ax.title.set_fontsize(9)
d_axes[0].legend(fontsize=7, loc='upper left')
d_figures.append(save_diagnostic_figure(d_fig, d_figure_dir, 'd5_distribusi_permutasi',
    f"Tiga ilustrasi memakai {len(d_signs)} partisi lengkap masing-masing dan toleransi sumber atol={d_config['atol']}, rtol={d_config['rtol']}. Jumlah ekstrem/p-value cocok baseline untuk semua contoh. Adjusted p-value tetap milik keluarga global {len(d_core)} pengujian; tidak ada uji LOPO atau penafsiran p sebagai probabilitas hipotesis nol benar."))
```
